# Imports and pull in data

```python
import pandas as pd

customers_df = pd.read_csv("../data/raw/customers.csv")
accounts_df = pd.read_csv("../data/raw/accounts.csv")
transactions_df = pd.read_csv("../data/raw/transactions.csv")
```

# Start exploration

```python
print(accounts_df.head())
print(accounts_df.describe())
print(accounts_df["customer_id"].isnull().sum())
```

# Continue checking

```python
print(accounts_df["credit_limit"].mean())
print(accounts_df["current_balance"].mean())
print(accounts_df["account_status"].value_counts())
```

# Additional Checks

```python
print(len(customers_df))
print(len(accounts_df))
print(len(transactions_df))

print(customers_df.isnull().sum())
print(accounts_df.isnull().sum())
print(transactions_df.isnull().sum())
```
